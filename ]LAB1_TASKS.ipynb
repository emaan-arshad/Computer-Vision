```python
Lab 01 Tasks Implementation
Part 1: Environment Setup & Validation
Python
# Task 1.1: Installation
# %pip install opencv-python
# %pip install matplotlib

# Task 1.2: Import & Validate
import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
print(f"OpenCV version: {cv2.__version__}")
Task 1: Object-Oriented Grocery Manager
Python
class GroceryManager:
    def __init__(self):
        self.inventory = {}

    def add_item(self, item, quantity, price):
        self.inventory[item] = {'quantity': quantity, 'price': price}

    def remove_item(self, item):
        if item in self.inventory:
            del self.inventory[item]
        else:
            print(f"Error: {item} does not exist in the list.")

    def view_list(self):
        for item, details in self.inventory.items():
            print(f"{item}: {details['quantity']} @ ${details['price']}")

    def calculate_total(self):
        return sum(details['quantity'] * details['price'] for details in self.inventory.values())
Task 2: Advanced Student Record System
Python
students = {
    'S1': {'Name': 'Alice', 'Major': 'CS', 'Grades': [85, 90, 92]},
    'S2': {'Name': 'Bob', 'Major': 'AI', 'Grades': [70, 75, 80]},
    'S3': {'Name': 'Charlie', 'Major': 'CS', 'Grades': [95, 98, 99]}
}

def highest_average(record_sys):
    best_student = ""
    max_avg = 0
    for sid, info in record_sys.items():
        avg = sum(info['Grades']) / len(info['Grades'])
        if avg > max_avg:
            max_avg = avg
            best_student = info['Name']
    return best_student

def search_major(record_sys, major):
    print(f"Students in {major}:")
    for info in record_sys.values():
        if info['Major'] == major:
            print(info['Name'])
Task 3: Safe Image Loading & RGB Visualization
Python
image_path = 'Sukuna.jpeg'
image = cv2.imread(image_path)

if image is None:
    print("Error: Image not found.")
else:
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(8,6))
    plt.axis('off')
    plt.title('King of Curses Ryomen Sukuna', fontsize=16, color='darkred', pad=15)
    plt.imshow(image_rgb)
    plt.show()
Task 4: Interactive Geometry & Blank Canvas
Python
canvas = np.zeros((800, 800, 3), dtype=np.uint8)
center = (400, 400)

# 5 concentric circles of alternating colors
colors = [(255,0,0), (0,255,0), (0,0,255), (255,255,0), (0,255,255)]
for i in range(5):
    radius = 350 - (i * 60)
    cv2.circle(canvas, center, radius, colors[i], -1)

# Bounding box around outermost circle (radius 350)
cv2.rectangle(canvas, (50, 50), (750, 750), (255, 255, 255), 3)

plt.imshow(canvas)
plt.title(f'Center Coordinates: {center}')
plt.axis('off')
plt.show()
Task 5: Advanced Blurring & NumPy ROI Extraction
Python
image = cv2.imread('Sukuna.jpeg')
if image is not None:
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    blurred = cv2.GaussianBlur(image_rgb, (25, 25), 0)

    h, w = image_rgb.shape[:2]
    cx, cy = w // 2, h // 2

    # 300x300 ROI from exact center
    roi_orig = image_rgb[cy-150:cy+150, cx-150:cx+150]
    roi_blur = blurred[cy-150:cy+150, cx-150:cx+150]

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(roi_orig)
    plt.title('Original ROI', fontsize=14, color='blue')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(roi_blur)
    plt.title('Blurred ROI', fontsize=14, color='red')
    plt.axis('off')
    plt.show()
Task 6: Alpha Blending & Typography
Python
image = cv2.imread('Sukuna.jpeg')
if image is not None:
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    h, w = image_rgb.shape[:2]

    overlay = image_rgb.copy()
    box_height = int(h * 0.20)
    cv2.rectangle(overlay, (0, h - box_height), (w, h), (0, 0, 150), -1)

    blended = cv2.addWeighted(overlay, 0.5, image_rgb, 0.5, 0)

    # Text over the transparent box
    text_y = int(h - (box_height / 2))
    cv2.putText(blended, "Stylized Caption", (50, text_y), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)

    plt.imshow(blended)
    plt.axis('off')
    plt.show()
Task 7: Matrix-Based Rotation & Adaptive Thresholding
Python
image = cv2.imread('document.jpg', cv2.IMREAD_GRAYSCALE)
if image is not None:
    _, global_thresh = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)
    adaptive_thresh = cv2.adaptiveThreshold(image, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)

    # Rotation Matrix scaling by 0.8
    h, w = adaptive_thresh.shape
    M = cv2.getRotationMatrix2D((w/2, h/2), 45, 0.8)
    rotated_adaptive = cv2.warpAffine(adaptive_thresh, M, (w, h))

    plt.figure(figsize=(15, 5))
    plt.subplot(1, 3, 1), plt.imshow(global_thresh, cmap='gray'), plt.title('Global Binary'), plt.axis('off')
    plt.subplot(1, 3, 2), plt.imshow(adaptive_thresh, cmap='gray'), plt.title('Adaptive'), plt.axis('off')
    plt.subplot(1, 3, 3), plt.imshow(rotated_adaptive, cmap='gray'), plt.title('Rotated Adaptive'), plt.axis('off')
    plt.show()
Task 8: Masking with Bitwise Logic
Python
img1 = cv2.imread('Sukuna.jpeg')
img2 = cv2.imread('Gojo.jpeg')

if img1 is not None and img2 is not None:
    img1 = cv2.resize(img1, (500, 500))
    img2 = cv2.resize(img2, (500, 500))

    mask = np.zeros((500, 500), dtype=np.uint8)
    cv2.circle(mask, (250, 250), 150, 255, -1)

    cutout_img1 = cv2.bitwise_and(img1, img1, mask=mask)

    mask_inv = cv2.bitwise_not(mask)
    cutout_img2 = cv2.bitwise_and(img2, img2, mask=mask_inv)

    combined = cv2.bitwise_or(cutout_img1, cutout_img2)
    combined_rgb = cv2.cvtColor(combined, cv2.COLOR_BGR2RGB)

    plt.imshow(combined_rgb)
    plt.axis('off')
    plt.show()
Task 9: Statistical Image Profiling with Pandas
Python
import pandas as pd

image = cv2.imread('Sukuna.jpeg')
if image is not None:
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Flatten the array and load into DataFrame
    flattened = image_rgb.reshape(-1, 3)
    df = pd.DataFrame(flattened, columns=['Red', 'Green', 'Blue'])

    print(df.describe())
```
